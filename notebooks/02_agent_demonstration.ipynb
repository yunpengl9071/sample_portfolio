# TrialSense AI - Agent Demonstration

This notebook demonstrates the multi-agent system using LangGraph.

```python
import sys
sys.path.append('..')

import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Check API keys are configured
if not os.getenv('OPENAI_API_KEY') and not os.getenv('ANTHROPIC_API_KEY'):
    print("⚠️ Warning: No API keys configured. Set OPENAI_API_KEY or ANTHROPIC_API_KEY in .env")
else:
    print("✓ API keys configured")
```

## 1. Initialize the Orchestrator

```python
from trialsense.agents.orchestrator import TrialSenseOrchestrator

# Initialize orchestrator
orchestrator = TrialSenseOrchestrator()

print("Orchestrator initialized with:")
print(f"  - Outcome Prediction Agent")
print(f"  - Site Matching Agent")
print(f"  - LangGraph Workflow")
```

## 2. Example 1: Outcome Prediction

```python
# Query for outcome prediction
query = "Predict the success probability for trial NCT04567890"

result = await orchestrator.arun(query)

print("="*80)
print("QUERY:", query)
print("="*80)
print()
print("RESPONSE:")
print(result['response'])
print()
print("="*80)
print("REASONING TRACE:")
for i, step in enumerate(result['reasoning_trace'], 1):
    print(f"  {i}. {step}")
```

## 3. Example 2: Site Matching

```python
# Query for site recommendations
query = "Recommend optimal sites for a Phase 3 lung cancer immunotherapy trial"

result = await orchestrator.arun(query)

print("="*80)
print("QUERY:", query)
print("="*80)
print()
print("RESPONSE:")
print(result['response'])
print()
print("Task Type:", result['task_type'])
```

## 4. Example 3: Comprehensive Analysis

```python
# Query for comprehensive analysis
query = "Provide comprehensive analysis for trial NCT03842513"

result = await orchestrator.arun(query)

print("="*80)
print("QUERY:", query)
print("="*80)
print()
print("RESPONSE:")
print(result['response'])
print()
print("="*80)
print("WORKFLOW:")
print(f"  Task Type: {result['task_type']}")
print(f"  Steps Executed: {len(result['reasoning_trace'])}")
```

## 5. Testing Individual Agents

```python
from trialsense.agents.outcome_agent import OutcomePredictionAgent

# Test outcome agent directly
outcome_agent = OutcomePredictionAgent(verbose=True)

result = await outcome_agent.arun(
    "What is the predicted outcome for a Phase 3 oncology trial with 300 patients?"
)

print(result['output'])
```

## 6. Visualizing the LangGraph Workflow

```python
# The LangGraph workflow can be visualized
# This shows the agent routing and state transitions

print("Workflow Nodes:")
print("  1. Router - Analyzes query and determines task type")
print("  2. Outcome Predictor - Runs ML prediction + evidence gathering")
print("  3. Site Matcher - Recommends optimal trial sites")
print("  4. Synthesizer - Combines results into final response")
print()
print("State Management:")
print("  - Query and task type")
print("  - NCT ID extraction")
print("  - Outcome prediction results")
print("  - Site recommendations")
print("  - Reasoning trace")
```
